Este cuaderno utiliza SOLO imágenes sintéticas: **no son observaciones NASA**. Demuestra cómo alinear (aquí ya alineadas), restar e identificar cambios.


```python
from pathlib import Path
import sys
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))
from backend.app.services.demo_service import make_demo_pair
from backend.app.analysis.image_difference import difference_image
from backend.app.analysis.motion_detection import detect_candidates
import matplotlib.pyplot as plt
import numpy as np
```

## 1. Simular dos épocas
El generador crea estrellas estáticas, una fuente variable y un objeto móvil artificial.


```python
a, b, meta = make_demo_pair(127.69444, -39.1776, 0, 2)
print(meta)
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
for axis, image, title in zip(ax, [a, b], ["Época A · demo", "Época B · demo"]):
    axis.imshow(image, origin="lower", cmap="magma", vmin=np.percentile(a, 5), vmax=np.percentile(a, 99.7))
    axis.set_title(title)
plt.show()
```

## 2. Calcular diferencias robustas
No todas las diferencias representan desplazamiento.


```python
diff, statistics = difference_image(a, b)
candidates = detect_candidates(diff, statistics["rms_robust"])
print("Ruido robusto:", statistics["rms_robust"])
for item in candidates:
    print(item["id"], item["classification"], "SNR ~", item["snr"], "desplazamiento(px)", item["shift_px"])
plt.figure(figsize=(7, 6))
plt.imshow(diff, origin="lower", cmap="coolwarm", vmin=-20, vmax=20)
plt.colorbar(label="Cambio (unidades arbitrarias)")
plt.title("B - A · residuales simulados")
plt.show()
```
